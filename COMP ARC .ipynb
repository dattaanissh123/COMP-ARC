```python
def ROR(binary, positions):
    print("\n--- Rotate Right Without Carry (ROR) ---")
    print("Original 8-bit number:", binary)

    for i in range(positions):
        last_bit = binary[-1]
        binary = last_bit + binary[:-1]
        print(f"Step {i + 1}: {binary}")

    print("Final 8-bit result:", binary)


def RCR(binary, positions, carry):
    print("\n--- Rotate Right Through Carry (RCR) ---")
    print("Original 8-bit number:", binary)
    print("Initial carry flag:", carry)

    for i in range(positions):
        last_bit = binary[-1]
        binary = str(carry) + binary[:-1]
        carry = int(last_bit)

        print(f"Step {i + 1}: {binary}, Carry = {carry}")

    print("Final 8-bit result:", binary)
    print("Final carry flag:", carry)


# (a) ROR: use the last 8 bits of the given number
ROR("10111101", 3)

# (b) RCR: use the last 8 bits and assume initial carry = 0
RCR("10100001", 3, 0)
def to_binary(number, bits=8):
    return format(number & ((1 << bits) - 1), f'0{bits}b')


def signed_value(binary):
    
    bits = len(binary)
    value = int(binary, 2)

    if value & (1 << (bits - 1)):
        value -= (1 << bits)

    return value


def perform_operation(a, b, operation):
    bits = 8
    minimum = -(1 << (bits - 1))
    maximum = (1 << (bits - 1)) - 1

    print("\nOperand A:", a, "=", to_binary(a))
    print("Operand B:", b, "=", to_binary(b))

    if operation == '+':
        result = a + b
        print("\nOperation: Addition")
    else:
        result = a - b
        print("\nOperation: Subtraction")

    # Display the result in 16-bit 2's complement format
    result_16 = to_binary(result, 16)

    # Display the result in 8-bit format
    result_8 = to_binary(result, 8)

    print("8-bit result:", result_8)
    print("16-bit result:", result_16)
    print("Decimal result:", result)

    # Check for 8-bit signed overflow
    overflow = result < minimum or result > maximum

    if overflow:
        print("8-bit Arithmetic Overflow: YES")
    else:
        print("8-bit Arithmetic Overflow: NO")


# a) +56 + (+29)
print("Expression (a): +56 + (+29)")
perform_operation(56, 29, '+')

# b) (-38) - (-23)
print("\nExpression (b): (-38) - (-23)")
perform_operation(-38, -23, '-')
```
